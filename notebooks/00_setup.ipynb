```python
# Sanity check on a toy dataset (fruit-fly lifespans, NOT lymphoma data)
from lifelines.datasets import load_waltons
from lifelines import KaplanMeierFitter
from lifelines.statistics import logrank_test
import matplotlib.pyplot as plt

df = load_waltons()
ax = plt.gca()
groups = {g: sub for g, sub in df.groupby('group')}
for g, sub in groups.items():
    KaplanMeierFitter().fit(sub['T'], sub['E'], label=g).plot_survival_function(ax=ax)

a, b = groups.values()
res = logrank_test(a['T'], b['T'], a['E'], b['E'])
print(f"Log-rank p = {res.p_value:.2e}")
plt.title('Sanity check: lifelines toy data'); plt.show()

```
